## Render a scene / Renderizar una escena

`%%manim` takes the same flags as the `manimgl` command. `-l` renders at 480p, which is plenty for a notebook. The video is written to `videos/` next to this notebook and shown below the cell.

`%%manim` acepta los mismos flags que el comando `manimgl`. `-l` renderiza a 480p, suficiente para un notebook. El video se escribe en `videos/` junto a este notebook y se muestra bajo la celda.

```python
%%manim -l HelloManimGL
class HelloManimGL(Scene):
    def construct(self):
        title = Text("ManimGL in Jupyter", font_size=72)
        square = Square(side_length=2, color=BLUE, fill_opacity=0.5)
        self.play(Write(title))
        self.play(title.animate.to_edge(UP), ShowCreation(square))
        self.play(Rotate(square, PI / 4))
        self.play(square.animate.set_fill(YELLOW).scale(1.5))
        self.wait()
```

## Last frame only / Solo el último fotograma

`-s` skips the animations and saves the final frame as a PNG. Handy for figures.

`-s` omite las animaciones y guarda el fotograma final como PNG. Útil para figuras.

```python
%%manim -s GraphFigure
class GraphFigure(Scene):
    def construct(self):
        axes = Axes(x_range=(-3, 3), y_range=(-1, 4), width=10, height=6)
        axes.add_coordinate_labels()
        graph = axes.get_graph(lambda x: x**2 / 2, color=GREEN)
        label = Tex(R"y = \frac{x^2}{2}", font_size=60).next_to(graph.get_end(), LEFT)
        self.add(axes, graph, label)
```

## Line magic and display size / Magia de línea y tamaño

Scenes defined earlier can be rendered again with the line form. `jupyter_config` controls how results are shown.

Las escenas ya definidas se pueden renderizar de nuevo con la forma de línea. `jupyter_config` controla cómo se muestran los resultados.

```python
jupyter_config.media_width = "100%"
%manim -m HelloManimGL
```

## Common flags / Flags habituales

| Flag | |
|---|---|
| `-l` `-m` `--hd` `--uhd` | 480p, 720p, 1080p, 4k |
| `-r 1280x720`, `--fps 60` | resolution and frame rate / resolución y fps |
| `-s` | last frame as PNG / último fotograma en PNG |
| `-i` | GIF |
| `-c "#112233"` | background color / color de fondo |
| `-n 2,5` | only animations 2 to 5 / solo animaciones 2 a 5 |
| `--prerun` | progress bar with a total / barra de progreso con total |

`%manim -h` lists them all / muestra todos.
